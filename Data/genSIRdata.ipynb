```python
import numpy as np
import pandas as pd

# 1. กำหนดพารามิเตอร์เริ่มต้น (เพื่อให้เกิด 1 Full Cycle ใน 365 วัน)
days = 365
N = 1000.0
beta = 0.15
gamma = 0.05

# กำหนดค่าเริ่มต้น S, I, R (เริ่มจากติดเชื้อแค่ 1 คน)
S = [999.0]
I = [1.0]
R = [0.0]

# 2. จำลองข้อมูล SIR ด้วยสมการพื้นฐาน
for t in range(days - 1):
    inf = beta * S[-1] * I[-1] / N
    rec = gamma * I[-1]
    
    S.append(S[-1] - inf)
    I.append(I[-1] + inf - rec)
    R.append(R[-1] + rec)

S = np.array(S)
I = np.array(I)
R = np.array(R)

# 3. ใส่ Noise (Gaussian Noise) เพื่อให้ดูเหมือนข้อมูลจริง
np.random.seed(42)
# Noise จะสวิงกว้างขึ้นเมื่อค่าจริงมีขนาดใหญ่ขึ้น
I_noisy = I + np.random.normal(0, 2 + I * 0.05, days)
R_noisy = R + np.random.normal(0, 5 + R * 0.02, days)

# ป้องกันไม่ให้ค่าที่ติด Noise มีค่าติดลบ และปัดเศษทศนิยม 1 ตำแหน่ง
I_noisy = np.maximum(0, I_noisy).round(1)
R_noisy = np.maximum(0, R_noisy).round(1)

# 4. สร้าง DataFrame
df = pd.DataFrame({
    "Day": np.arange(days),
    "Infected_Data": I_noisy,
    "Recovered_Data": R_noisy
})

# 5. แทรก "ข้อมูลขยะ" บางจุดเพื่อ Stress Test ระบบ (จำลองการกรอกข้อมูลผิด)
df.loc[120, "Infected_Data"] = "error"
df.loc[250, "Recovered_Data"] = np.nan
df.loc[340, "Infected_Data"] = "-"

# 6. บันทึกเป็นไฟล์ CSV
csv_filename = "sir_data_365_days.csv"
df.to_csv(csv_filename, index=False)

print(f"✅ สร้างไฟล์ {csv_filename} สำเร็จ!")
print(f"📊 จำนวนข้อมูล: {len(df)} แถว (ครอบคลุม 1 Full Cycle พร้อม Noise)")
```
